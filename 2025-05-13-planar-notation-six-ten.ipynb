```
bqBlock[{line10_, line6_}, offset_] := {
    Table[Text[Style[line10[[i]], FontSize -> 15], {- i + 1, 0} + {-1/2, 1/2} + offset], {i, Length[line10]}],
    Table[Text[Style[line6[[i]], FontSize -> 15], {0, i} + {-1/2, 1/2} + offset], {i, Length[line6]}],
    Transparent, EdgeForm[Black],
    Polygon[# + offset & /@ {{0, 0}, {-2, 0}, {-2, 1}, {-1, 1}, {0, 1}, {0, 0}}]
}
```

```
bqGraphics[li_] := Graphics[
    Table[bqBlock[li[[i]], {-i , i}], {i, Length[li]}],
    PlotRange -> All,
    AspectRatio -> Automatic,
    ImageSize -> Small
]
```
