# 離散TSP

巡回セールスマン問題を格子点上の点群で考える

```
(* 8x8の格子点を生成 *)
points = Flatten[Table[{i, j}, {i, 0, 7}, {j, 0, 7}], 1];

(* 異なる4つの点をランダムに選択 *)
selectedPoints = RandomSample[points, 4];

(* 距離関数を定義 *)
distance[{x1_, y1_}, {x2_, y2_}] := Sqrt[(x2 - x1)^2 + (y2 - y1)^2];

(* 全ての点の順列を生成 *)
permutations = Permutations[selectedPoints];

(* 各順列の総距離を計算 *)
totalDistances = Total[distance @@@ Partition[#, 2, 1]] & /@ permutations;

(* 最短経路を算出 *)
minDistance = Min[totalDistances];

(* 結果を表示 *)
{selectedPoints, minDistance}
```
