```
Module[{alpha, r, theta},
  alphaMax = 1;
  alphaMin = 0;
  k = 1;

  alpha[n_] := alphaMin + (alphaMax - alphaMin) / (1 + k Log[n]);
  r[n_] := n^alpha[n];
  theta[n_] := 2 Pi Log[2, n];

  Graphics[
    Table[
      Disk[AngleVector[{r[i], theta[i]}], 1 / 128],
      {i, 1, 1024}
    ],
    ImageSize -> 256
  ]
]
```

```
Module[{alphaMax, alphaMin, k, alpha, r, theta, alphaP},
  alphaMax = 1;
  alphaMin = 0;
  k = 1;

  alpha[n_] := alphaMin + (alphaMax - alphaMin) / (1 + k Log[n]);
  r[n_] := n^alpha[n];
  theta[n_] := 2 Pi Log[2, n];

  alphaP[n_] := - (alphaMax - alphaMin) k / (n (1 + k Log[n])^2);
  size[n_] := r[n] Sqrt[(alphaP[n] Log[n] + alpha[n] / n)^2 + (2 Pi / (n Log[2]))^2] / 128;

  Graphics[
    Table[
      Disk[AngleVector[{r[i], theta[i]}], size[i]],
      {i, 1, 4096}
    ],
    ImageSize -> 256
  ]
]
```
