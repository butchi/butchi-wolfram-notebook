```
balancedDecimal[n_Integer] :=
   Module[{digits = {}, r, d, m = n},
      If[m == 0, Return[{0}]];
      While[m != 0,
         r = Mod[m, 10];
         If[r > 4,
            d = r - 10;
            m = (m - d)/10,
            d = r;
            m = (m - r)/10
         ];
      AppendTo[digits, d]
   ];
   Reverse[digits]
]

Grid[Prepend[
     Table[{n, PadLeft[balancedDecimal[n], 5]}, {n, 0, 15}], 
     {"n", "Balanced Decimal Representation"}
]]
```
