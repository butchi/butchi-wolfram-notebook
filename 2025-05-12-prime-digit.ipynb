```
flatFactorInteger[n_] := Join @@ (ConstantArray @@@ FactorInteger[n])
```

```
integerTree[0] := 0
integerTree[1] := 1
integerTree[2] := 2
integerTree[3] := 3
integerTree[n_Integer?PrimeQ] := succ[integerTree[n - 1]]
integerTree[n_Integer?CompositeQ] := times @@ (integerTree /@ flatFactorInteger[n])
```
