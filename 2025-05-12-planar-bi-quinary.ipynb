```
bqBlock[{line2_, line5_}, offset_] := {
    Table[Text[Style[line2[[i]], FontSize -> 15], {- i + 1, 0} + {-1/2, 1/2} + offset], {i, Length[line2]}],
    Table[Text[Style[line5[[i]], FontSize -> 15], {0, i} + {-1/2, 1/2} + offset], {i, Length[line5]}],
    Transparent, EdgeForm[Black],
    Polygon[# + offset & /@ {{0, 0}, {-3, 0}, {-3, 1}, {-1, 1}, {-1, 2}, {0, 2}, {0, 0}}]
}
```

```
bqGraphics[li_] := Graphics[
    Table[bqBlock[li[[i]], {-i , i}], {i, Length[li]}],
    PlotRange -> All,
    AspectRatio -> Automatic,
    ImageSize -> Small
]
```
