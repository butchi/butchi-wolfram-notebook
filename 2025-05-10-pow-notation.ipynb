# 記数法

 * 素数による記数法: 素数だったら1を減らして素因数分解 → n番目の素数番目の素数…の記数法は以前にやった
 * 二五十進法（二番煎じ？）: 左シフトは2倍、上シフトは5倍

```
integerPrimes[0] := 0;
integerPrimes[1] := 1;
integerPrimes[2] := power[2, 1];
integerPrimes[n_Integer?PrimeQ] := plus[1, integerPrimes[n - 1]]
integerPrimes[n_Integer] := Inner[power, Sequence @@ Transpose[FactorInteger[n]], times];
```

```
Grid[Table[{i, integerPrimes[i]}, {i, 0, 99}]]
```

```
Block[{
    power = Power[ToString[#1], #2] &,
    times = Times,
    plus = Plus
}, Grid[Table[{i, integerPrimes[i]}, {i, 99}]]]
```

```
Block[{
    power = Power[ToString[#1], #2] &,
    times = Times,
    plus = Plus
}, Grid[Table[{i, integerPrimes[i]}, {i, Prime[Range[15]]}]]]
```

```
Block[{
    power = Power[ToString[#1], #2] &,
    times = Times,
    plus = Plus
}, Grid[Table[{i, integerPrimes[i]}, {i, (Prime[Range[2, 15]] - 1)/2 }]]]
```

```
integerTree[0] := 0;
integerTree[1] := 1;
integerTree[2] := 2;
integerTree[n_Integer?PrimeQ] := plus[1, integerTree[n - 1]]
integerTree[n_Integer] := # & /@ Inner[power, Sequence @@ Transpose[{integerTree[#[[1]]], #[[2]]} & /@ FactorInteger[n]], times];
```

```
integerTree[97]
```

```
Grid[Table[{i, integerTree[i]}, {i, 15}]]
```

```
Table[TreeForm[integerTree[i]], {i, 15}]
```

```
factorPrimePlus1[n_] := FactorInteger[n + 1];
factorPrimeMinus1[n_] := FactorInteger[n - 1];
```

```
factorPrimePlus1[17]
factorPrimeMinus1[17]
```

```
expand2[expr_] := z (expr - PolynomialMod[expr, 2]) / 2 + PolynomialMod[expr, 2]
expand5[expr_] := s (expr - PolynomialMod[expr, 5]) / 5 + PolynomialMod[expr, 5]
```

```
Grid[Partition[Table[Grid[Reverse /@ Reverse @ PadRight[CoefficientList[FixedPoint[Expand[expand2[expand5[#]]] &, i, 99], {s, z}], {4, 4}], Frame -> True], {i, 100}], 10]]
```

```
Grid[Reverse /@ Reverse @ PadRight[CoefficientList[FixedPoint[Expand[expand2[expand5[#]]] &, 123456, 99], {s, z}]]]
```
