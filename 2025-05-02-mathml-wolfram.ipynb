```
equal = Equal;
sup = Power;
plus = Plus;
times = Times;
frac = Divide;

equal[
  sup[times[plus["a", "b"]], 2],
  plus[
    sup["c", 2],
    times[4, times[frac[1, 2], "a", "b"]]
  ]
]

equal[
  plus[
    sup["a", 2],
    times[2, "a", "b"],
    sup["b", 2]
  ],
  plus[
    sup["c", 2],
    times[2, "a", "b"]
  ]
]

equal[
  plus[
    sup["a", 2],
    sup["b", 2]
  ],
  sup["c", 2]
]
```
