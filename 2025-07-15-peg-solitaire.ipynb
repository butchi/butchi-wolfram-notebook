```
1 + x == x^2
x^2 + x == 1

x^m == x^(m + 1) + x^(m + 2)
x^n == x^(n - 1) + x^(n - 2)

x^(m + 1) + x^(m + 2) - x^m == 0
x^(n - 1) + x^(n - 2) - x^n == 0

x^(m + 1) + x^(m + 2) - x^m == x^(n - 1) + x^(n - 2) - x^n
```
